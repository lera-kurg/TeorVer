Тема 4. Парадоксы в теории вероятностей
Рассмотреть какой-нибудь парадокс. Смоделировать его решение в Python.


Участник телевизионной игры видит **три закрытые двери**. За одной находится автомобиль, за двумя другими — козы.

1. Участник выбирает одну дверь.
2. Ведущий, **зная**, где автомобиль, открывает другую дверь, за которой находится коза.
3. Ведущий предлагает участнику **оставить выбор** или **сменить дверь** на единственную оставшуюся закрытую.

Какая стратегия увеличивает вероятность выиграть автомобиль?

**Важные условия:** ведущий всегда открывает дверь с козой, не открывает дверь участника и всегда предлагает смену. 


При первом выборе участник угадывает расположение автомобиля с вероятностью $1/3$, а ошибается с вероятностью $2/3$.

- **Не менять дверь:** победа возможна только тогда, когда первоначальный выбор верный. Поэтому $P(\text{выигрыш без смены})=\frac{1}{3}$.
- **Менять дверь:** победа возможна только тогда, когда первоначальный выбор неверный. Ведущий исключает одну дверь с козой, и за оставшейся закрытой дверью оказывается автомобиль. Поэтому $P(\text{выигрыш со сменой})=\frac{2}{3}$.

Следовательно, смена двери **удваивает вероятность выигрыша**: примерно с 33,3% до 66,7%.

| Первоначальный выбор | Вероятность | Без смены | Со сменой |
|---|---:|---|---|
| Автомобиль | $1/3$ | Выигрыш | Проигрыш |
| Коза | $2/3$ | Проигрыш | Выигрыш |

```python
import random
import matplotlib.pyplot as plt

n = 10000

wins_stay = 0
wins_switch = 0

st = []
sw = []

for i in range(1, n + 1):
    # Случайное расположение автомобиля
    car = random.randint(0, 2)

    # Первоначальный выбор участника
    choice = random.randint(0, 2)

    # Ведущий открывает дверь с козой
    possible = [
        door for door in range(3)
        if door != car and door != choice
    ]
    opened = random.choice(possible)

    # Дверь, на которую можно поменять выбор
    new_choice = [
        door for door in range(3)
        if door != choice and door != opened
    ][0]

    # Проверяем обе стратегии
    if choice == car:
        wins_stay += 1

    if new_choice == car:
        wins_switch += 1

    # Сохраняем частоты выигрышей
    st.append(wins_stay / i)
    sw.append(wins_switch / i)

print("Количество игр:", n)
print("Без смены двери:", round(wins_stay / n, 4))
print("Со сменой двери:", round(wins_switch / n, 4))
```
